```python
import pandas as pd
import numpy as np

df = pd.read_csv("./Datasets/stations_prepared.csv")
```

```python
# Create mean values for plotting
df["temp_mean"] = (df["temp_min"] + df["temp_max"]) / 2
df["do_mean"]   = (df["do_min"] + df["do_max"]) / 2
df["ph_mean"]   = (df["ph_min"] + df["ph_max"]) / 2
df["cond_mean"] = (df["conductivity_min"] + df["conductivity_max"]) / 2
df["bod_mean"]  = (df["bod_min"] + df["bod_max"]) / 2
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df.index, df["ph_mean"])
plt.xlabel("Monitoring Station Index")
plt.ylabel("Mean pH")
plt.title("Variation of pH Across Monitoring Stations")
# plt.show()
plt.savefig("Graphs/Variation_of_pH_Across_Monitoring_Stations.png")
```

```python
# DO
plt.figure()
plt.plot(df.index, df["do_mean"])
plt.xlabel("Monitoring Station Index")
plt.ylabel("Dissolved Oxygen (mg/L)")
plt.title("DO Variation Across Monitoring Stations")
# plt.show()
plt.savefig("Graphs/DO_Variation_Across_Monitoring_Stations.png")

# Turbidity (using max as worst-case)
plt.figure()
plt.plot(df.index, df["bod_max"])
plt.xlabel("Monitoring Station Index")
plt.ylabel("BOD (mg/L)")
plt.title("BOD Variation Across Monitoring Stations")
# plt.show()
plt.savefig("Graphs/BOD_Variation_Across_Monitoring_Stations.png")

# Conductivity
plt.figure()
plt.plot(df.index, df["cond_mean"])
plt.xlabel("Monitoring Station Index")
plt.ylabel("Conductivity (µS/cm)")
plt.title("Conductivity Variation Across Monitoring Stations")
# plt.show()
plt.savefig("Graphs/Conductivity_Variation_Across_Monitoring_Stations.png")
```

```python
plt.figure()
plt.hist(df["ph_mean"], bins=20)
plt.xlabel("Mean pH")
plt.ylabel("Frequency")
plt.title("Distribution of Mean pH Values")
# plt.show()
plt.savefig("Graphs/Distribution_of_Mean_pH_Values.png")

plt.figure()
plt.hist(df["do_mean"], bins=20)
plt.xlabel("Dissolved Oxygen (mg/L)")
plt.ylabel("Frequency")
plt.title("Distribution of DO Values")
# plt.show()
plt.savefig("Graphs/Distribution_of_DO_Values.png")

df["do_mean"].plot(kind="kde", title="KDE of DO Values")
plt.xlabel("Dissolved Oxygen (mg/L)")
# plt.show()
plt.savefig("Graphs/KDE_of_DO_Values.png")
```

```python
plt.figure()
plt.scatter(df["bod_mean"], df["do_mean"])
plt.xlabel("BOD (mg/L)")
plt.ylabel("Dissolved Oxygen (mg/L)")
plt.title("Relationship between BOD and DO")

# Regression line
m, b = np.polyfit(df["bod_mean"], df["do_mean"], 1)
plt.plot(df["bod_mean"], m*df["bod_mean"] + b)

# plt.show()
plt.savefig("Graphs/Relationship_between_BOD_and_DO.png")

plt.figure()
plt.scatter(df["cond_mean"], df["ph_mean"])
plt.xlabel("Conductivity (µS/cm)")
plt.ylabel("pH")
plt.title("Conductivity vs pH")

m, b = np.polyfit(df["cond_mean"], df["ph_mean"], 1)
plt.plot(df["cond_mean"], m*df["cond_mean"] + b)

# plt.show()
plt.savefig("Graphs/Conductivity_vs_pH.png")
```

```python
plt.figure()
df["wqi_category"].value_counts().plot(kind="bar")
plt.xlabel("WQI Category")
plt.ylabel("Number of Stations")
plt.title("Distribution of Water Quality Index Categories")
# plt.show()
plt.savefig("Graphs/Distribution_of_Water_Quality_Index_Categories.png")
```
